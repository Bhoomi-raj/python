```python
students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    
    "Name": [
        "Bhoomi", "Ayush", "Parth", "Payal",
        "Sonam", "Sanjeev", "Rahul", "Priya"
    ],
    
    "Age": [18, 19, 20, 21, 22, 20, 19, 18],
    
    "Course": [
        "CSE", "IT", "ECE", "ME",
        "CSE", "IT", "ECE", "CSE"
    ],
    
    "Salary": [
        25000, 30000, 28000, 26000,
        35000, 40000, 32000, 29000
    ]
})
fees = pd.DataFrame({
    "Student_ID": [101, 102, 103, 105, 107, 108],
    
    "Fee_Status": [
        "Paid", "Pending", "Paid",
        "Paid", "Pending", "Paid"
    ],
    
    "Scholarship": [
        5000, 0, 3000,
        4000, 0, 2000
    ]
})
```

```python
students
```

```python
fees
```

```python
pd.merge(students,fees)
```

```python
pd.merge(students,fees,on = "Student_ID",how = "right")
```

```python
pd.merge(students,fees,on = "Student_ID",how = "outer")
```
