```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flight = sns.load_dataset('flights')
flight
```

```python

tips = sns.load_dataset('tips')
tips
```

```python
tipscorr = tips[['total_bill','tip','size']]
sns.heatmap(tipscorr.corr(),annot = True)
```

```python
sns.clustermap(tipscorr.corr(),annot = True)
```

```python
flightpivot= flight.pivot_table(values='passengers',index='month',columns='year')
flightpivot
```

```python
sns.heatmap(flightpivot)
```

```python
sns.lmplot(x ='total_bill',y= 'tip',data = tips)
```
