```python
import numpy as np
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(1,10,10)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("first plot")
plt.xlabel("time")
plt.ylabel("velocity")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x,x)
plt.subplot(2,2,4)
plt.plot(y,y)
```

```python
fig = plt.figure(figsize = (20,8),dpi=100)
axis1 = fig.add_axes([.1,.1,.8,.8])
x = np.linspace(1,10,10)
y = np.linspace(1,5,10)
axis1.plot(x,y)



axis2 = fig.add_axes([.2,.6,.3,.3])
axis2.plot(x,y)
plt.show()
```
